```python
import polars as pl
```

```python
# Đọc file Parquet
df_items = pl.read_parquet("items.parquet")
df_trans = pl.read_parquet("transactions-2025-12.parquet")
```

```python
df_items = df_items.filter(pl.col("category_l1").str.contains("(?i)Tã"))
```

```python
df_items = df_items.filter(pl.col("size") != "Không xác định")
# Loại bỏ các sản phẩm không còn bán
df_items = df_items.filter(pl.col("sale_status") == 1)
# Giữ lại những dòng có size KHÁC '0-3M'
df_items = df_items.filter(pl.col("size") != "0-3M")
#df_items = df_items.filter(pl.col("size") != "1Y")
```

```python
print(df_items["size"].unique().to_list())
```

```python
# Lọc các dòng có size đúng bằng '0-3M'
df_03m = df_items.filter(pl.col("size") == "2Y")

# Hiển thị kết quả
with pl.Config(fmt_str_lengths=10000):
    
    print(df_03m)
```

```python
df_items = df_items.with_columns(
    pl.col("size")
    .str.extract(r"(XXL|XL|S|M|L)", 1) # Tìm các chữ cái này trong chuỗi
    .alias("size")
)

# Hiển thị kết quả để kiểm tra
print(df_items)
```
